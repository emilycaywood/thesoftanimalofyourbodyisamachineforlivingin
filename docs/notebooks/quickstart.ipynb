# CALFLAB from a notebook

Start the lab first (`.\calflab.ps1 lab`). This notebook talks to the same server and project as the web app, so edits made here appear there live.

Units: mm, g, degrees.

```python
from calflab.client import Client

lab = Client()  # http://127.0.0.1:8000
lab.health()
```

```python
scene = lab.scene()
scene["mass"], scene["extents"], scene["com"]
```

## Sweep a gene and simulate
Each `simulate()` is recorded in the project's registry with its genome, seed and code version.

```python
rows = []
for shank in (150, 170, 190, 210):
    lab.set_genes(shank_length=shank)
    run = lab.simulate()
    m = run["metrics"]
    rows.append((shank, run["id"], round(m["speed_mps"], 3), round(m["cost_of_transport"], 2), m["fell"]))
rows
```

## Explicit override, analysis, undo

```python
lab.override("leg.fl.shank", "length", 200, name="long front-left shank")
margins = lab.analysis("torque_margin")["rows"]
[(r["joint"], r["required_peak_nm"], r["available_nm"], r["margin"]) for r in margins[:6]]
```

```python
lab.undo()  # shared undo: also undoes in the web app
```
